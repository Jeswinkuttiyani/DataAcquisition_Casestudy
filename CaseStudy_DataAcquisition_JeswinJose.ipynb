```python
import pandas as pd
import sqlite3
```

```python
#TASK 1
filepath="/content/drive/MyDrive/Data Science/Case Study/DataAcquisitionCasestudy_B10/population.csv"
csv1=pd.read_csv(filepath)
print(csv1.head(5),"\n")
print("Shape:",csv1.shape,"\n")
print(csv1.dtypes)
```

```python
#Task 2
excel1 = pd.read_excel("/content/drive/MyDrive/Data Science/Case Study/DataAcquisitionCasestudy_B10/gdp.xlsx")
print("Column Names:")
print(excel1.columns)

print("\nData Types:")
print(excel1.dtypes)
```

```python
#Task 3
filepath=("/content/drive/MyDrive/Data Science/Case Study/DataAcquisitionCasestudy_B10/internet_users.json")
json1=pd.read_json(filepath)
print(json1)
```

```python
#Task 4
xml1 = pd.read_xml("/content/drive/MyDrive/Data Science/Case Study/DataAcquisitionCasestudy_B10/literacy_rate.xml")
print(xml1)
```

```python
#Task 5
# Population Dataset
print("===== Population Dataset =====")
print("Rows:", csv1.shape[0])
print("Columns:", csv1.shape[1])
print("Column Names:", csv1.columns.tolist())
print("\nData Types:")
print(csv1.dtypes)
print("\nMissing Values:")
print(csv1.isnull().sum())
print("\nDuplicate Records:", csv1.duplicated().sum())


# GDP Dataset
print("\n===== GDP Dataset =====")
print("Rows:", excel1.shape[0])
print("Columns:", excel1.shape[1])
print("Column Names:", excel1.columns.tolist())
print("\nData Types:")
print(excel1.dtypes)
print("\nMissing Values:")
print(excel1.isnull().sum())
print("\nDuplicate Records:", excel1.duplicated().sum())


# Internet Users Dataset
print("\n===== Internet Users Dataset =====")
print("Rows:", json1.shape[0])
print("Columns:", json1.shape[1])
print("Column Names:", json1.columns.tolist())
print("\nData Types:")
print(json1.dtypes)
print("\nMissing Values:")
print(json1.isnull().sum())
print("\nDuplicate Records:", json1.duplicated().sum())


# Literacy Rate Dataset
print("\n===== Literacy Rate Dataset =====")
print("Rows:", xml1.shape[0])
print("Columns:", xml1.shape[1])
print("Column Names:", xml1.columns.tolist())
print("\nData Types:")
print(xml1.dtypes)
print("\nMissing Values:")
print(xml1.isnull().sum())
print("\nDuplicate Records:", xml1.duplicated().sum())
```

```python
# Task 6

excel1.rename(columns={'Country': 'country'}, inplace=True)

xml1.rename(columns={'name': 'country'}, inplace=True)
print("Population:", csv1.columns.tolist())
print("GDP:", excel1.columns.tolist())
print("Internet Users:", json1.columns.tolist())
print("Literacy Rate:", xml1.columns.tolist())
```

```python
# Task 7 - Convert columns to numeric data types

csv1['population'] = pd.to_numeric(csv1['population'])
excel1['GDP'] = pd.to_numeric(excel1['GDP'])
json1['internet_users'] = pd.to_numeric(json1['internet_users'])
xml1['literacy_rate'] = pd.to_numeric(xml1['literacy_rate'])
print(csv1.dtypes)
print(excel1.dtypes)
print(json1.dtypes)
print(xml1.dtypes)
```

```python
# Task 8
merged_df = pd.merge(csv1, excel1, on='country', how='inner')
merged_df = pd.merge(merged_df, json1, on='country', how='inner')
merged_df = pd.merge(merged_df, xml1, on='country', how='inner')
print(merged_df)
```

```python
# Task 9 - Create calculated columns
merged_df['Internet Penetration (%)'] = (
    merged_df['internet_users'] / merged_df['population'] * 100
).round(2)

merged_df['GDP Per Capita'] = (
    merged_df['GDP'] / merged_df['population']
).round(2)

print(merged_df)
```

```python
# Task 10
conn = sqlite3.connect('country_stats.db')
merged_df.to_sql('country_statistics', conn, if_exists='replace', index=False)

query_df = pd.read_sql_query('SELECT * FROM country_statistics LIMIT 5;', conn)
display(query_df)
```

```python
#Task 11
# Retrieve the first five records from the SQLite table..
result = pd.read_sql("SELECT * FROM country_statistics LIMIT 5", conn)
print(result)
```

# **DATA ANALYSIS USING PANDAS**

```python
# 1 List the Top 5 countries by GDP.
print("Top 5 countries by GDP.")
print(merged_df.nlargest(5,"GDP")[["country","GDP"]]);

# 2 List the Top 5 countries by population.
print("\nTop 5 countries by population.")
print(merged_df.nlargest(5,"population")[["country","population"]]);

# 3 Calculate the average literacy rate.
print("\nAverage literacy rate.")
print(merged_df["literacy_rate"].mean())

# 4 List all countries having a literacy rate greater than 90%.
print("\nLiteracy rate greater than 90%.")
print(merged_df[merged_df["literacy_rate"]>90][["country","literacy_rate"]])

# 5 List all countries having an internet penetration greater than 70%.
print("\nInternet penetration greater than 70%.")
print(merged_df[merged_df["Internet Penetration (%)"]>70][["country","Internet Penetration (%)"]])
```

```python
# 6 List the Top 5 countries by GDP per capita.
print("Countries by GDP per capita.");
print(merged_df.nlargest(5,"GDP Per Capita")[["country","GDP Per Capita"]]);

# 7 Calculate the total population represented in the dataset.
print("\nTotal Population.");
print(merged_df["population"].sum());

# 8 Calculate the average internet penetration across all countries.
print("\nAverage Internet Penetration");
print(merged_df["Internet Penetration (%)"].mean());

# 9 Identify the country having the highest number of internet users.
print("\nHighest number of internet users.");
print(merged_df.loc[merged_df["internet_users"].idxmax(),["country","internet_users"]]);

# # 10 Find the correlation between Literacy Rate and Internet Penetration.
print("\nCorrelation between Literacy Rate and Internet Penetration.");
print(merged_df["literacy_rate"].corr(merged_df["Internet Penetration (%)"]));
```

# **Data Analysis Using SQL**

```python
# 11 Display the Top 5 countries by GDP.

q = """select country, GDP from country_statistics order by GDP DESC LIMIT 5"""
print("\nTop 5 countries by GDP.")
print(pd.read_sql(q, conn))

# 12 Find the average literacy rate.
print("\nAverage literacy rate.")
q = """select AVG(literacy_rate) as average_literacy_rate from country_statistics"""
print(pd.read_sql(q, conn))

# 13 Display all countries having literacy rate above 90%.
print("\nCountries having literacy rate above 90%.")
q = """select country, literacy_rate from country_statistics where literacy_rate>90"""
print(pd.read_sql(q, conn))

# 14 Find the Top 5 countries with the highest internet penetration.
print("\nTop 5 countries with the highest internet penetration.")
q = """select country, "Internet Penetration (%)" from country_statistics
ORDER BY "Internet Penetration (%)" DESC LIMIT 5"""
print(pd.read_sql(q, conn))

# 15 Calculate the total population..
print("\nTotal population.")
q = """select sum(population) as total_population from country_statistics"""
print(pd.read_sql(q, conn))

# 16 Find the country with the highest GDP per capita.
print("\nCountry with the highest GDP per capita.")
q = """select country, "GDP Per Capita" from country_statistics order by "GDP Per Capita" DESC LIMIT 1"""
print(pd.read_sql(q, conn))

# 17 Export the final merged dataset as a CSV file.
merged_df.to_csv("final_country_statistics.csv", index=False)
print("\nFinal CSV exported.")

# 18 Sort the dataset in descending order of GDP per Capita and display the Top 10
result = merged_df.sort_values("GDP Per Capita", ascending=False)
print(result.head(10)[["country", "GDP Per Capita"]])
```
